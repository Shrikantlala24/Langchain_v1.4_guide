```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
def LLM_config():
    load_dotenv()

    llm = ChatGoogleGenerativeAI(
        model="gemini-3.1-flash-lite",
        temperature=1
    )

    return llm

llm = LLM_config()

response = llm.invoke("who are you")
response
```
